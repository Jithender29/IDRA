Import Libraries and Load Dataset


```python
import pandas as pd
import numpy as np
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 120)
file_path = 'Day11_Messy_Company_Employee_Dataset.csv'
df = pd.read_csv(file_path)
print('Dataset loaded successfully.')
print('Shape:', df.shape)
df.head()
```

Inspect the Dataset Before Cleaning


```python
print('Column names:')
print(df.columns.tolist())

print('\nData types:')
df.dtypes.to_frame('Data Type')

print('\nDataset information:')
df.info()

print('\nDescriptive statistics:')
df.describe(include='all')
```

Identify Missing Values


```python
missing_before = df.isnull().sum().sort_values(ascending=False)
print('Missing values in each column:')
missing_before.to_frame('Missing Values')

print('Total missing cells:', int(df.isnull().sum().sum()))
```

Identify Duplicate Records


```python
duplicate_count = int(df.duplicated().sum())
print('Duplicate records:', duplicate_count)

if duplicate_count > 0:
    df[df.duplicated(keep=False)].sort_values(df.columns.tolist())
```

Detect Inconsistent Text Entries


```python
text_columns = df.select_dtypes(include='object').columns.tolist()

for col in text_columns:
    print(f'\nUnique values in {col}:')
    print(df[col].value_counts(dropna=False).to_string())
```

Clean Whitespace and Blank Entries


```python
for col in text_columns:
    df[col] = df[col].apply(lambda x: x.strip() if isinstance(x, str) else x)
    df[col] = df[col].replace(r'^\s*$', np.nan, regex=True)

print('Whitespace and blank entries cleaned.')
```

Standardize Inconsistent Categorical Values


```python
standard_maps = {
    'Department': {
        'hr': 'HR', 'human resources': 'HR',
        'it': 'IT', 'information technology': 'IT',
        'finance': 'Finance', 'fin': 'Finance',
        'marketing': 'Marketing', 'mkt': 'Marketing',
        'sales': 'Sales', 'operations': 'Operations', 'ops': 'Operations'
    },
    'Gender': {
        'm': 'Male', 'male': 'Male', 'man': 'Male',
        'f': 'Female', 'female': 'Female', 'woman': 'Female'
    },
    'Work_Mode': {
        'wfh': 'Remote', 'remote': 'Remote', 'work from home': 'Remote',
        'office': 'Onsite', 'on-site': 'Onsite', 'onsite': 'Onsite',
        'hybrid': 'Hybrid'
    },
    'City': {
        'new delhi': 'Delhi', 'delhi': 'Delhi',
        'bangalore': 'Bengaluru', 'bengaluru': 'Bengaluru',
        'bombay': 'Mumbai', 'mumbai': 'Mumbai',
        'calcutta': 'Kolkata', 'kolkata': 'Kolkata',
        'madras': 'Chennai', 'chennai': 'Chennai',
        'hyderabad': 'Hyderabad', 'pune': 'Pune'
    }
}

for col, mapping in standard_maps.items():
    if col in df.columns:
        df[col] = df[col].apply(
            lambda x: mapping.get(str(x).strip().lower(), x) if pd.notna(x) else x
        )

if 'Employee_Name' in df.columns:
    df['Employee_Name'] = df['Employee_Name'].apply(
        lambda x: ' '.join(str(x).split()).title() if pd.notna(x) else x
    )

if 'Job_Title' in df.columns:
    df['Job_Title'] = df['Job_Title'].apply(
        lambda x: ' '.join(str(x).split()).title() if pd.notna(x) else x
    )

print('Categorical values standardized.')
```

Correct Data Types


```python
df['Employee_ID'] = df['Employee_ID'].astype('string').str.strip()
df['Joining_Date'] = pd.to_datetime(df['Joining_Date'], errors='coerce')

numeric_columns = ['Age', 'Annual_Salary', 'Experience_Years', 'Performance_Score']
for col in numeric_columns:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

print('Corrected data types:')
df.dtypes.to_frame('Data Type')
```

Handle Missing Values


```python
numeric_impute = ['Age', 'Annual_Salary', 'Experience_Years', 'Performance_Score']
categorical_impute = ['Department', 'Gender', 'City', 'Work_Mode']

for col in numeric_impute:
    if col in df.columns and df[col].isnull().any():
        df[col] = df[col].fillna(df[col].median())

for col in categorical_impute:
    if col in df.columns and df[col].isnull().any():
        df[col] = df[col].fillna(df[col].mode(dropna=True)[0])

print('Missing values after imputation:')
df.isnull().sum().to_frame('Missing Values')
```

Remove Duplicate Records


```python
before_duplicates = df.duplicated().sum()
df = df.drop_duplicates().copy()
after_duplicates = df.duplicated().sum()

print('Duplicates before:', before_duplicates)
print('Duplicates removed:', before_duplicates - after_duplicates)
print('Duplicates after:', after_duplicates)
```

Final Type and Format Cleanup


```python
if 'Age' in df.columns:
    df['Age'] = df['Age'].round().astype('Int64')
if 'Experience_Years' in df.columns:
    df['Experience_Years'] = df['Experience_Years'].round().astype('Int64')
if 'Performance_Score' in df.columns:
    df['Performance_Score'] = df['Performance_Score'].round().astype('Int64')
if 'Annual_Salary' in df.columns:
    df['Annual_Salary'] = df['Annual_Salary'].round(2)

print(df.dtypes.to_frame('Final Data Type'))
df.head()
```

Before vs After Cleaning


```python
print('BEFORE CLEANING')
print('Rows:', 157)
print('Columns:', 12)
print('Missing cells:', 32)
print('Duplicate rows:', 7)

print('\nAFTER CLEANING')
print('Rows:', len(df))
print('Columns:', len(df.columns))
print('Missing cells:', int(df.isnull().sum().sum()))
print('Duplicate rows:', int(df.duplicated().sum()))

comparison = pd.DataFrame({
    'Before': [157, 12, 32, 7],
    'After': [len(df), len(df.columns), int(df.isnull().sum().sum()), int(df.duplicated().sum())]
}, index=['Rows', 'Columns', 'Missing Cells', 'Duplicate Rows'])
print(comparison)
```

Verify the Cleaned Dataset


```python
print('Total missing cells:', int(df.isnull().sum().sum()))
print('Total duplicate rows:', int(df.duplicated().sum()))
print('\nFinal data types:')
print(df.dtypes.to_frame('Data Type'))
print('\nFinal dataset sample:')
display(df.head(10))

assert df.isnull().sum().sum() == 0
assert df.duplicated().sum() == 0
print('Verification passed: no missing values or duplicate records remain.')
```

Export Cleaned Dataset


```python
output_file = 'Day11_Cleaned_Company_Employee_Dataset.csv'
df.to_csv(output_file, index=False, date_format='%Y-%m-%d')
print(f'Cleaned dataset exported successfully to: {output_file}')
```
