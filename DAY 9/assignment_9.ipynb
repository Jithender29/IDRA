Import Pandas and Load the Datasets

```python
import pandas as pd

orders = pd.read_csv("Day9_Orders.csv")
customers = pd.read_csv("Day9_Customers.csv")
products = pd.read_csv("Day9_Products.csv")

print("Orders dataset:")
print(orders.head())

print("\nCustomers dataset:")
print(customers.head())

print("\nProducts dataset:")
print(products.head())
```

Check Dataset Shapes

```python
print("Orders dataset shape:")
print(orders.shape)

print("\nCustomers dataset shape:")
print(customers.shape)

print("\nProducts dataset shape:")
print(products.shape)
```

Check Column Names

```python
print("Orders columns:")
print(orders.columns)

print("\nCustomers columns:")
print(customers.columns)

print("\nProducts columns:")
print(products.columns)
```

Check Missing Values

```python
print("Missing values in Orders:")
print(orders.isnull().sum())

print("\nMissing values in Customers:")
print(customers.isnull().sum())

print("\nMissing values in Products:")
print(products.isnull().sum())
```

Merge Orders with Customers

```python
order_customer = pd.merge(
    orders,
    customers,
    on="Customer_ID",
    how="left"
)

print("Orders combined with customer information:")
print(order_customer.head())
```

Merge with Products

```python
processed_data = pd.merge(
    order_customer,
    products,
    on="Product_ID",
    how="left"
)

print("Final merged dataset:")
print(processed_data.head())
```

Check the Merged Dataset

```python
print("Shape of the merged dataset:")
print(processed_data.shape)

print("\nColumns in the merged dataset:")
print(processed_data.columns)
```

Demonstrate concat()

```python
sample_orders = orders.head(5)
sample_orders_2 = orders.tail(5)

combined_orders = pd.concat(
    [sample_orders, sample_orders_2],
    ignore_index=True
)

print("Orders combined using concat:")
print(combined_orders)
```

Convert Order Date to DateTime

```python
processed_data["Order_Date"] = pd.to_datetime(processed_data["Order_Date"])

print("Order date after converting to DateTime:")
print(processed_data["Order_Date"].head())
```

```python
processed_data["Order_Month"] = processed_data["Order_Date"].dt.month

print("Month extracted from order date:")
print(processed_data[["Order_Date", "Order_Month"]].head())
```

```python
processed_data["Month_Name"] = processed_data["Order_Date"].dt.month_name()

print("Month name for each order:")
print(processed_data[["Order_Date", "Month_Name"]].head())
```

```python
processed_data["Order_Day"] = processed_data["Order_Date"].dt.day

print("Day extracted from order date:")
print(processed_data[["Order_Date", "Order_Day"]].head())
```

```python
processed_data["Day_of_Week"] = processed_data["Order_Date"].dt.day_name()

print("Day of the week for each order:")
print(processed_data[["Order_Date", "Day_of_Week"]].head())
```

Create Total Sales Using apply()

```python
processed_data["Total_Sales"] = processed_data.apply(
    lambda row: row["Quantity"] * row["Unit_Price"],
    axis=1
)

print("Total sales calculated for each order:")
print(processed_data[
    ["Order_ID", "Product_Name", "Quantity", "Unit_Price", "Total_Sales"]
].head(10))
```

```python
processed_data["Price_Category"] = processed_data["Unit_Price"].apply(
    lambda x: "High" if x >= 2000 else "Low"
)

print("Price category of products:")
print(processed_data[
    ["Product_Name", "Unit_Price", "Price_Category"]
].head(10))
```

```python
processed_data["Quantity_Type"] = processed_data["Quantity"].apply(
    lambda x: "Bulk Order" if x >= 3 else "Normal Order"
)

print("Quantity type of each order:")
print(processed_data[
    ["Order_ID", "Quantity", "Quantity_Type"]
].head(10))
```

Create the Final Clean Dataset

```python
final_data = processed_data[
    [
        "Order_ID", "Order_Date", "Order_Month", "Month_Name",
        "Order_Day", "Day_of_Week", "Customer_ID", "Customer_Name",
        "City", "Region", "Membership_Type", "Product_ID",
        "Product_Name", "Category", "Brand", "Unit_Price",
        "Quantity", "Total_Sales", "Payment_Method", "Order_Status",
        "Price_Category", "Quantity_Type"
    ]
]

print("Clean processed dataset:")
print(final_data.head())
```

Sort the Processed Dataset

```python
print("Orders sorted by total sales:")
print(final_data.sort_values("Total_Sales", ascending=False).head(10))
```

Check the Final Dataset

```python
print("Final dataset shape:")
print(final_data.shape)

print("\nFinal dataset information:")
final_data.info()

print("\nMissing values in final dataset:")
print(final_data.isnull().sum())
```

Basic Analysis

```python
print("Total sales:")
print(final_data["Total_Sales"].sum())

print("\nAverage sales:")
print(round(final_data["Total_Sales"].mean(), 2))

print("\nTotal quantity sold:")
print(final_data["Quantity"].sum())
```

```python
print("Sales by category:")
print(final_data.groupby("Category")["Total_Sales"].sum().sort_values(ascending=False))
```

```python
print("Sales by city:")
print(final_data.groupby("City")["Total_Sales"].sum().sort_values(ascending=False))
```

```python
print("Sales by membership type:")
print(final_data.groupby("Membership_Type")["Total_Sales"].sum().sort_values(ascending=False))
```

Export the Final Dataset

```python
final_data.to_csv("Processed_Ecommerce_Dataset.csv", index=False)
print("Processed dataset has been exported successfully.")
```

Final Output

```python
print("Final processed e-commerce dataset:")
print(final_data.head(10))

print("\nNumber of rows:", len(final_data))
print("Number of columns:", len(final_data.columns))
print("Processed dataset is ready for submission.")
```
