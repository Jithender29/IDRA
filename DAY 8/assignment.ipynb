```python
import pandas as pd
df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")
df.head()
```

```python
print("Shape of the dataset:")
print(df.shape)
```

```python
print("Column names:")
print(df.columns)
```

```python
print("Information about the dataset:")
df.info()
```

```python
print("Statistical summary of the dataset:")
print(df.describe())
```

```python
print("Checking for missing values:")
print(df.isnull().sum())
```

Data Selection

```python
print("Product column:")
print(df["Product"].head())
```

```python
print("Selected columns from the dataset:")
print(df[["Order_ID", "Customer", "Product", "Quantity", "Total_Sales"]].head())
```

```python
print("First row of the dataset:")
print(df.iloc[0])
```

```python
print("First 5 orders with selected details:")
print(df.loc[0:4, ["Order_ID", "Customer", "City", "Total_Sales"]])
```

Filtering

```python
print("Orders having sales greater than 5000:")
print(df[df["Total_Sales"] > 5000])
```

```python
print("Orders from Hyderabad:")
print(df[df["City"] == "Hyderabad"])
```

```python
print("Orders paid using UPI:")
print(df[df["Payment_Method"] == "UPI"])
```

```python
print("Sports orders where quantity is 3 or more:")
print(df[(df["Category"] == "Sports") & (df["Quantity"] >= 3)])
```

```python
print("Orders having a discount of 10% or more:")
print(df[df["Discount_Percent"] >= 10])
```

Sorting

```python
print("Top 10 orders based on sales:")
print(df.sort_values("Total_Sales", ascending=False).head(10))
```

```python
print("10 orders with the lowest sales:")
print(df.sort_values("Total_Sales").head(10))
```

```python
print("Top 10 orders based on quantity:")
print(df.sort_values("Quantity", ascending=False).head(10))
```

Overall Sales Statistics

```python
print("Total sales:", df["Total_Sales"].sum())
print("Average sales:", round(df["Total_Sales"].mean(), 2))
print("Highest sale:", df["Total_Sales"].max())
print("Lowest sale:", df["Total_Sales"].min())
print("Total quantity sold:", df["Quantity"].sum())
print("Number of orders:", df["Order_ID"].nunique())
```

Sales by Category

```python
print("Total sales for each category:")
print(df.groupby("Category")["Total_Sales"].sum().sort_values(ascending=False))
```

```python
print("Category-wise sales, average sales, quantity and number of orders:")
print(df.groupby("Category").agg({
    "Total_Sales": ["sum", "mean", "max", "min"],
    "Quantity": "sum",
    "Order_ID": "count"
}).sort_values(("Total_Sales", "sum"), ascending=False))
```

Sales by City

```python
print("Total sales for each city:")
print(df.groupby("City")["Total_Sales"].sum().sort_values(ascending=False))
```

```python
print("City-wise sales, average sales, quantity and number of orders:")
print(df.groupby("City").agg({
    "Total_Sales": ["sum", "mean"],
    "Quantity": "sum",
    "Order_ID": "count"
}).sort_values(("Total_Sales", "sum"), ascending=False))
```

Sales by Product

```python
print("Top 10 products based on total sales:")
print(df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False).head(10))
```

```python
print("Product-wise sales, average sales, quantity and number of orders:")
print(df.groupby("Product").agg({
    "Total_Sales": ["sum", "mean"],
    "Quantity": "sum",
    "Order_ID": "count"
}).sort_values(("Total_Sales", "sum"), ascending=False).head(10))
```

Sales by Payment Method

```python
print("Total sales for each payment method:")
print(df.groupby("Payment_Method")["Total_Sales"].sum().sort_values(ascending=False))
```

```python
print("Payment method-wise sales, average sales, quantity and orders:")
print(df.groupby("Payment_Method").agg({
    "Total_Sales": ["sum", "mean"],
    "Quantity": "sum",
    "Order_ID": "count"
}))
```

Sales by Region

```python
print("Total sales for each region:")
print(df.groupby("Region")["Total_Sales"].sum().sort_values(ascending=False))
```

```python
print("Region-wise sales, average sales, quantity and number of orders:")
print(df.groupby("Region").agg({
    "Total_Sales": ["sum", "mean"],
    "Quantity": "sum",
    "Order_ID": "count"
}).sort_values(("Total_Sales", "sum"), ascending=False))
```

Highest and Lowest Sales

```python
print("Order with the highest sales:")
print(df.loc[df["Total_Sales"].idxmax()])
```

```python
print("Order with the lowest sales:")
print(df.loc[df["Total_Sales"].idxmin()])
```

Top Performing Products

```python
print("Top 5 products based on total sales:")
print(df.groupby("Product")["Total_Sales"].sum().sort_values(ascending=False).head(5))
```

```python
print("Categories ranked by total sales:")
print(df.groupby("Category")["Total_Sales"].sum().sort_values(ascending=False))
```

```python
print("Sports orders paid using UPI with sales above 5000:")

result = df[
    (df["Category"] == "Sports") &
    (df["Payment_Method"] == "UPI") &
    (df["Total_Sales"] > 5000)
]

print(result)
```

```python
total_sales = df["Total_Sales"].sum()
average_sales = df["Total_Sales"].mean()
total_quantity = df["Quantity"].sum()

top_category = df.groupby("Category")["Total_Sales"].sum().idxmax()
top_product = df.groupby("Product")["Total_Sales"].sum().idxmax()
top_city = df.groupby("City")["Total_Sales"].sum().idxmax()
top_payment = df.groupby("Payment_Method")["Total_Sales"].sum().idxmax()

highest_order = df.loc[df["Total_Sales"].idxmax()]
lowest_order = df.loc[df["Total_Sales"].idxmin()]

print("Meaningful Observations:")
print()
print("1. Total sales generated:", round(total_sales, 2))
print("2. Average sales per order:", round(average_sales, 2))
print("3. Total quantity of products sold:", total_quantity)
print("4. Category with the highest sales:", top_category)
print("5. Product with the highest sales:", top_product)
print("6. City with the highest sales:", top_city)
print("7. Payment method with the highest sales:", top_payment)
print("8. Highest sales order:", highest_order["Order_ID"], "with sales of", highest_order["Total_Sales"])
print("9. Lowest sales order:", lowest_order["Order_ID"], "with sales of", lowest_order["Total_Sales"])
```
