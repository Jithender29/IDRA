Import Libraries and Load Dataset

```python
import pandas as pd
import numpy as np
from IPython.display import display
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 120)
file_path = "Day10_Flight_Operations_Dataset.csv"
df = pd.read_csv(file_path)
print("Dataset loaded successfully")
print("Shape:", df.shape)
```

Understand the Dataset

```python
display(df.head())
display(df.tail())
print("Columns:", df.columns.tolist())
print("Shape:", df.shape)
display(df.dtypes.to_frame("Data Type"))
df.info()
```

Data Quality Checks

```python
print("Missing values:")
display(df.isnull().sum().sort_values(ascending=False).to_frame("Missing Values"))
print("Duplicate rows:", df.duplicated().sum())
print("Unique values:")
display(df.nunique().sort_values().to_frame("Unique Values"))
```

Descriptive Statistics

```python
display(df.describe().T)
display(df.describe(include="all").T)
```

Selecting Relevant Data

```python
numeric_data = df.select_dtypes(include="number")
categorical_data = df.select_dtypes(exclude="number")
print("Numerical data:")
display(numeric_data.head())
print("Categorical data:")
display(categorical_data.head())
```

Filtering Data

```python
if len(numeric_data.columns):
    metric = numeric_data.columns[0]
    threshold = df[metric].median()
    filtered = df[df[metric] >= threshold]
    print(f"Records where {metric} >= median ({threshold:.2f}):")
    display(filtered.head(10))
```

Sorting Records

```python
if len(numeric_data.columns):
    metric = numeric_data.columns[0]
    sorted_data = df.sort_values(metric, ascending=False)
    print(f"Top records sorted by {metric}:")
    display(sorted_data.head(10))
```

Grouping and Aggregation

```python
group_col = 'Airline'
if group_col:
    print("Records by group:")
    display(df[group_col].value_counts().to_frame("Flight Count"))

    if len(numeric_data.columns):
        metric = numeric_data.columns[0]
        grouped = df.groupby(group_col)[metric].agg(["count","mean","min","max"])
        display(grouped.sort_values("mean", ascending=False))
```

Data Transformation

```python
if len(numeric_data.columns):
    metric = numeric_data.columns[0]
    median_value = df[metric].median()
    df[f"{metric}_Level"] = df[metric].apply(
        lambda x: "High" if pd.notna(x) and x >= median_value else "Low"
    )
    display(df[[metric, f"{metric}_Level"]].head(10))
```

Date/Time Analysis

```python
date_columns = [c for c in df.columns if any(k in c.lower() for k in ["date", "time"])]
print("Detected date/time columns:", date_columns)
for c in date_columns:
    parsed = pd.to_datetime(df[c], errors="coerce")
    if parsed.notna().any():
        parsed_col = c + "_Parsed"
        df[parsed_col] = parsed
        print(f"Converted: {c}")
        display(df[[c, parsed_col]].head())
```

```python
airline_col = 'Airline'
status_col = 'Flight_Status'
origin_col = 'Origin'
dest_col = 'Destination'
delay_col = 'Delay_Minutes'
duration_col = None
distance_col = None
if airline_col:
    print("1. Which airline/carrier has the most flights?")
    display(df[airline_col].value_counts().head(10).to_frame("Flight Count"))
if status_col:
    print("2. What are the most common flight statuses?")
    display(df[status_col].value_counts().to_frame("Count"))
if origin_col:
    print("3. Which origins are busiest?")
    display(df[origin_col].value_counts().head(10).to_frame("Flight Count"))
if dest_col:
    print("4. Which destinations are busiest?")
    display(df[dest_col].value_counts().head(10).to_frame("Flight Count"))
for label, col in [("Delay", delay_col), ("Duration", duration_col), ("Distance", distance_col)]:
    if col:
        values = pd.to_numeric(df[col], errors="coerce")
        print(f"Highest recorded {label} values:")
        display(df.loc[values.nlargest(10).index, [col]])
```

```python
airline_col = 'Airline'
status_col = 'Flight_Status'
metric_col = 'Delay_Minutes'

if airline_col and status_col:
    print("Airline/carrier vs status:")
    display(pd.crosstab(df[airline_col], df[status_col]))
if airline_col and metric_col:
    comparison = (
        df.groupby(airline_col)[metric_col]
          .agg(["count","mean","median","min","max"])
          .sort_values("mean", ascending=False)
    )
    print("Metric statistics by airline/carrier:", metric_col)
    display(comparison)
```

```python
observations = []
observations.append(f"1. The dataset contains {len(df):,} flight records and {df.shape[1]} columns.")
if 'Airline':
    x = df['Airline'].value_counts()
    if len(x):
        observations.append(f"2. The most represented airline/carrier is {x.index[0]} with {x.iloc[0]:,} records.")
if 'Flight_Status':
    x = df['Flight_Status'].value_counts()
    if len(x):
        observations.append(f"3. The most common flight status is {x.index[0]} with {x.iloc[0]:,} records.")
if 'Delay_Minutes':
    x = pd.to_numeric(df['Delay_Minutes'], errors="coerce").dropna()
    if len(x):
        observations.append(f"4. The average recorded delay is {x.mean():.2f}, while the maximum is {x.max():.2f}.")
if 'Origin':
    x = df['Origin'].value_counts()
    if len(x):
        observations.append(f"5. The busiest origin/departure location is {x.index[0]} with {x.iloc[0]:,} flights.")
if 'Destination':
    x = df['Destination'].value_counts()
    if len(x):
        observations.append(f"6. The most frequent destination/arrival location is {x.index[0]} with {x.iloc[0]:,} flights.")
if None:
    x = pd.to_numeric(df[None], errors="coerce").dropna()
    if len(x):
        observations.append(f"7. The average flight duration is {x.mean():.2f} in the dataset's recorded unit.")
if None:
    x = pd.to_numeric(df[None], errors="coerce").dropna()
    if len(x):
        observations.append(f"8. The average flight distance is {x.mean():.2f} in the dataset's recorded unit.")
for observation in observations[:8]:
    print(observation)
```
